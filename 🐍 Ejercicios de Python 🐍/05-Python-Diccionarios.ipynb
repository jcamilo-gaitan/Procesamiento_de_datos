<h2 id="Dic">Diccionarios</h2>

<h3 id="content">¿Qué son los diccionarios?</h3>

Un diccionario esta conformado por llaves y valores. Un diccionario puede compararse a una lista. En vez de índices numéricos como los de las listas, los diccionarios tienen llaves. Estas llaves son usadas para acceder a los valores dentro de los diccionarios.

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/DictsList.png" width="650" />

Un ejemplo de Diccionario <code>Dict</code>:

```python
# Crear el diccionario

Dict = {"key1": 1, "key2": "2", "key3": [3, 3, 3], "key4": (4, 4, 4), ('key5'): 5, (0, 1): 6}

Dict

#las llaves pueden ser varios tipos de datos:
```

Las llaves pueden ser cadenas:

```python
# Acceder a un valor con la llave 'key4'

Dict["key4"]
```

Una llave sólo puede ser cualquier tipo de objeto inmutable, pudiendo ser una tupla:

```python
# Acceder a un valor mediante una llave

Dict[(0, 1)]
```

Los diccionarios también pueden ser apuntados por varias variables, como en las listas , conjuntos o tuplas.

```python
dicci={1:'si',2:'no'}
dicci2=dicci
dicci2[2]='si'
dicci
#dicci se ha modificado porque igualamos ambos diccionarios y aplicamos una operacion sobre una de las dos variables

```

Cada llave esta separada de su respectivo valor mediante dos puntos "<code>:</code>". Las comas separan los elementos y todo el diccionario esta encerrado usando llaves. Un diccionario vacío se representa escribiendo solo dos llaves, de esta forma "<code>{}</code>".

```python
# Crear un ejemplo de diccionario

release_year_dict = {"Thriller": "1982", "Back in Black": "1980", \
                    "The Dark Side of the Moon": "1973", "The Bodyguard": "1992", \
                    "Bat Out of Hell": "1977", "Their Greatest Hits (1971-1975)": "1976", \
                    "Saturday Night Fever": "1977", "Rumours": "1977"}
release_year_dict
```

Resumiendo, al igual que una lista, un diccionario guarda una secuencia de elementos. Cada contiene una llave y su valor correspondiente. Los diccionarios se crean mediante dos llaves y estas contendrán las llaves y valores separados por una coma.

A cada llave solo puede corresponderle un valor, sin embargo, múltiples llaves pueden tener ese mismo valor. Las llaves solo pueden ser cadenas, números o tuplas, pero los valores pueden ser de cualquier tipo.

Es de ayuda visualizar un diccionario como una tabla, como se muestra en la imagen. La primera columna representa las llaves, la segunda los valores.

<h3 id="key">Llaves</h3>

Los valores se pueden obtener mediante nombres:

```python
# Acceder al valor con la llave 'Thriller'

release_year_dict['Thriller']
```

Ahora vamos a recuperar las llaves del diccionario mediante el método <code>release_year_dict()</code>:

```python
# Obtener todas las llaves del diccionario

release_year_dict.keys() # Esto no es un objeto list,set o tuple
```

Se puede obtener los valores mediante el método <code>values()</code>:

```python
# Obtener todos los valores del diccionario

release_year_dict.values() # Esto no es un objeto list,set o tuple
```

Podemos agregar una entrada:

```python
# Agregar un valor con su llave en el diccionario

release_year_dict['Graduation'] = '2007'
release_year_dict
```

Tambien podemos borrar entradas:

```python
# Borrar entradas mediante la llave

del(release_year_dict['Thriller'])
del(release_year_dict['Graduation'])
release_year_dict
```

Podemos verificar si existe un elemento en el diccionario:

```python
# Verificar si existe la llave en el diccionario

'The Bodyguard' in release_year_dict
```

```python
# Verificar si existe un valor en el diccionario (no llave)
1999 in release_year_dict.values()
```

También puedes filtrar un diccionario por condiciones especificas:

```python
#filtramos los discos que salieron desde el 80'
filtrados = {k:v for k,v in release_year_dict.items() if  int(v) >= 1980}
filtrados
```

Tambien existen los diccionarios anidados:

```python
#este dict contiene parejas clave valor: inicialmente las claves son cadenas pero los valores pueden ser otros dict.
datos = {
    "user1": {
        "id": 1,
        "profile": {
            "name": "Ana",
            "age": 20
        }
    },
    "user2": {
        "id": 2,
        "profile": {
            "name": "Juan",
            "age": 32
        }
    }
}
```

Considera el siguiente diccionario:

```python

estudiantes = {
    "Ana": {
        "edad": 20,
        "carrera": "Ingeniería",
        "notas": {
            "matematicas": 4.5,
            "fisica": 4.0,
            "programacion": 5.0
        }
    },

    "Luis": {
        "edad": 22,
        "carrera": "Derecho",
        "notas": {
            "matematicas": 3.2,
            "fisica": 2.8,
            "programacion": 4.1
        }
    },

    "Juan": {
        "edad": 21,
        "carrera": "Ciencia de Datos",
        "notas": {
            "matematicas": 4.8,
            "fisica": 4.6,
            "programacion": 4.9
        }
    }
}
```

a) Halla la nota que saco Ana en programación

```python
estudiantes['Ana']['notas']['programacion']
```

b)Halla la carrera de Luis

```python
estudiantes['Luis']['carrera']
```

c) Calcula el promedio de cada estudiante y crea un diccionario con el nombre del estudiantte y su promedio

```python
promedios = {
    n: sum(d["notas"].values()) / len(d["notas"])
    for n, d in estudiantes.items()
}
promedios
```

d) Genera un diccionario con los estudiantes con promedio mayor a 4.0

```python
filt = {k:v for k,v in promedios.items() if v> 4.0}
filt
```

## Conclusiones

Al igual que las listas o los conjuntos, varias variables pueden apuntar a un mismo diccionario, antes de hacer operaciones con este tipo de dato es importante tener eso en cuenta.

Se pueden diferenciar los diccionarios por que siempre mantienen iuna estructura clave - valor, donde la clave es irrepetible y debe ser un objeto inmutable (tuplas,cadenas o enteros)

Esta estructura permite acceder a cualquier tipo de dato desde una llave, lo que puede facilitar la programacion orientada a objetos mediante la abstracción que se puede generar con los dict. Además su estructura es similar al formato JSON.
