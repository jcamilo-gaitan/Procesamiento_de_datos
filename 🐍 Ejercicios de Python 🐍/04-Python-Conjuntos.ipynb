<h3 id="content">Contenido de un Conjunto</h3>

Un conjunto es una colección de objetos únicos en Python. Un conjunto se define usando llaves de apertura y cierre <b>{}</b>. Python removerá automáticamente elementos duplicados.

```python
# Crear un conjunto

set1 = {"pop", "rock", "soul", "hard rock", "rock", "R&B", "rock", "disco"}
set1
```

El proceso de mapeado se muestra en la siguiente figura:

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/SetsUnique.png" width="1100" />

También se puede crear un conjunto a partir de una lista:

```python
# Convertir una lista en un conjunto

album_list = [ "Michael Jackson", "Thriller", 1982, "00:42:19", \
              "Pop, Rock, R&B", 46.0, 65, "30-Nov-82", None, 10.0]
album_set = set(album_list)
album_set
```

Los conjuntos también pueden ser apuntados por varias variables, como en las listas.

```python
setX={"hola","chao"}
a=setX
a.add("adios")
print(setX)

```

Vamos a crear un conjunto de géneros musicales:

```python
# Convertir una lista en un conjunto

music_genres = set(["pop", "pop", "rock", "folk rock", "hard rock", "soul", \
                    "progressive rock", "soft rock", "R&B", "disco"])
music_genres
#en este caso, pop quedará en el conjunto 1 sola vez, porque este no permite duplicados
```

Como pudiste observar, los set no mantienen orden de inserción:

```python
sinorden={"hola","chao"}
print(sinorden)
sinorden={"dia","tarde"}
sinorden.add("noche")
print(sinorden)

#De hecho, no mantienen un orden claro, estos se ordenan segun códigos hash generados a partir del valor del elemento del conjunto.
```

<h3 id="op">Operaciones con Conjuntos</h3>

Vamos a revisar las operaciones con conjuntos que nos ayudaran a modificarlo. Considera el conjunto <b>A</b>:

```python
# Conjunto de ejemplo

A = set(["Thriller", "Back in Black", "AC/DC"])
A
```

Podemos agregar un elemento al conjunto mediante el método <code>add()</code>:

```python
# Agregar un elemento al conjunto

A.add("NSYNC")
A
```

Si agregamos el mismo elemento dos veces nada cambiara debido a que no puede haber elemento duplicados en un conjunto:

```python
# Trata de agregar un elemento duplicado al conjunto

A.add("NSYNC")
A
```

Podemos borrar un elemento del conjunto mediante el método <code>remove</code>:

```python
# Remover un elemento del conjunto

A.remove("NSYNC")
A
```

Podemos verificar si existe un elemento especifico dentro del conjunto con el comando <code>in</code>:

```python
# Verificar si existe un elemento en el conjunto

"AC/DC" in A
```

```python
album_set1.symmetric_difference(album_set2)
#Nótese que este resultado contiene album_set1.difference(album_set2) y album_set2.difference(album_set1)
# ¡ Esta operación retorna lo que esta en un conjunto o en el otro, pero no en ambos !
```

La <code>**unión**</code> es la suma de todos los elementos de ambos conjuntos, se representa pintando los dos círculos:

<img src = "https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/SetsUnion.png" width="650" />

La unión esta dada por:

```python
# Encontrar la unión de los dos conjuntos

album_set1.union(album_set2)
```

Se puede averiguar si un conjunto es un súper conjunto o un sub conjunto de otro, respectivamente, de la siguiente forma:

```python
# Revisar si album_set1 es un super conjunto de album_set2

set(album_set1).issuperset(album_set2)

#el conjunto album_set1 no contiene todos los elementos de album_set2
```

```python
# Revisar si album_set1 es un subconjunto de album_set2

set(album_set2).issubset(album_set1)

#el conjunto album_set1 completo no está en el conjunto album_set2
```

Este es un ejemplo donde <code>issubset()</code> y <code>issuperset()</code> regresan un valor verdadero:

```python
# Revisar si es un sub conjunto

set({"Back in Black", "AC/DC"}).issubset(album_set1)
#el conjunto set({"Back in Black", "AC/DC"} existe dentro del conjunto album_set1
```

```python
# Revisar si es un super conjunto

album_set1.issuperset({"Back in Black", "AC/DC"})
# album_set1 contiene todos los elementos de {"Back in Black", "AC/DC"}
```

Recuerda que :  
A ⊆ B → (A está contenido en B)   
es equivalente a

B ⊇ A → (B contiene a A)

## Conclusiones

Al igual que las listas, varias variables pueden apuntar a un mismo conjunto, antes de hacer operaciones es importante tener esto en cuenta.

Los conjuntos tienen una caracteristica especial que no tienen listas, tuplas  y otras estructuras: no permiten duplicados, esto facilita hacer operaciones matemáticas entre conjuntos.

Además, suele ser mas fácil buscar elementos dado que se usa una tabla hash por defecto.

A pesar de su sencillez, la forma de aprender estos procedimientos correctamente es la práctica, se sugiere al lector realizar varios ejercicios relacionados.

Algunos de los aprendizajes claves fueron:


*   Diferencia (Símetrica y no simetrica)
*   Unión
*   Sub/Super conjuntos    
*   Construcción de conjuntos

```python
# Conjuntos
agradecimiento = {"Gracias", "por"}
accion = {"revisar", "este", "cuaderno"}
valores = {"con", "dedicación", "y", "compromiso"}
proposito = {"en", "tu", "aprendizaje", "📘✨"}


mensaje_set = agradecimiento | accion | valores | proposito  #Uniones

# Orden lógico con lista
orden = [
    "Gracias", "por", "revisar", "este", "cuaderno",
    "con", "dedicación", "y", "compromiso",
    "en", "tu", "aprendizaje", "📘✨"
]

# Filtrar según el set  (los set se ordenan de acuerdo una tabla hash, no guardan orden de inserción)
mensaje_final = [p for p in orden if p in mensaje_set]

print(" ".join(mensaje_final))
```
